## 4. Select columns
Often we only need a few columns at a time.

```python
df[["Player", "Team", "League", "HR", "AVG", "OPS"]].head(10)
```

## 5. Filter rows
Filtering answers questions like: **Which hitters meet a condition?**

```python
# Example: players with at least 30 home runs
power_hitters = df[df["HR"] >= 30]
power_hitters[["Player", "Team", "HR", "OPS"]]
```

```python
elite_ops = df[df["OPS"] > .900]
elite_ops[["Player", "Team", "OPS"]].sort_values("OPS", ascending=False)
```

```python
nl_power = df[(df["League"] == "NL") & (df["HR"] >= 25)]
nl_power[["Player", "Team", "HR", "OPS"]]
```

## 6. Sort and find leaders
Sorting is one of the fastest ways to turn raw data into an answer.

```python
df.sort_values("HR", ascending=False)[["Player", "Team", "HR"]].head(10)
```

```python
df.nlargest(10, "OPS")[["Player", "Team", "OPS"]]
```

## 7. Summarize with groupby
`groupby()` splits data into groups, calculates something for each group, and combines the results.

```python
# Average stats by league *within this selected 50-player sample*
league_summary = df.groupby("League")[["HR", "RBI", "AVG", "OBP", "SLG", "OPS"]].mean()
league_summary.round(3)
```

```python
df.groupby("Position")["OPS"].mean().sort_values(ascending=False).round(3)
```

## 8. Plotly - turn the analysis into something visual
Plotly charts are interactive: hover, zoom, and inspect individual points.

```python
# Chart 1: Top 10 home run hitters
# We sort ascending here so the longest horizontal bar appears at the top.
top_hr = df.nlargest(10, "HR").sort_values("HR")

fig = px.bar(
    top_hr,
    x="HR",
    y="Player",
    orientation="h",
    hover_data=["Team", "League", "OPS"],
    title="Home Runs in the Workshop Dataset"
)
fig.show()
```

```python
# Chart 2: On-base skill vs. power
fig = px.scatter(
    df,
    x="OBP",
    y="SLG",
    color="League",
    size="HR",
    hover_name="Player",
    hover_data=["Team", "HR", "AVG", "OPS"],
    title="On-Base Percentage vs. Slugging Percentage"
)
fig.show()
```
